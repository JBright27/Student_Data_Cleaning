```python
cleaned_df = df.copy()
```

```python
cleaned_df.columns = (
    cleaned_df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)
```

```python
for column in cleaned_df.select_dtypes(include="object").columns:
    cleaned_df[column] = cleaned_df[column].str.strip()
```

# Student Performance Data Cleaning

## Dataset
The dataset contains student information and academic performance records.

## Data Cleaning Steps
The following steps were performed:

1. Loaded the original CSV dataset using Pandas.
2. Inspected the dataset structure, columns, data types and statistical summary.
3. Checked for missing values.
4. Checked for duplicate records.
5. Checked categorical and numerical values for consistency.
6. Standardized column names by removing unnecessary spaces, converting them to lowercase and replacing spaces with underscores.
7. Removed unnecessary leading and trailing spaces from text values.
8. Validated numerical values such as age, grades and number of failures.
9. Verified that the cleaned dataset retained 649 rows and 33 columns.
10. Saved the cleaned dataset as `cleaned_student_por.csv`.

## Data Quality Result
The dataset contained no missing values and no duplicate rows. The numerical values were within reasonable ranges, so no valid records were unnecessarily removed.

The cleaned dataset was saved as `cleaned_student_por.csv`.
